```python
from google.colab import files
uploaded = files.upload()
INPUT = next(iter(uploaded))
print('Input:', INPUT)
```

```python
PROMPT = '''A realistic cinematic video. The person in the reference image performs natural full-body dance movements with smooth coordinated arms and legs, realistic body mechanics, stable face and clothing, subtle camera movement, detailed lighting, natural background motion, high temporal consistency.'''
print(PROMPT)
```

```python
import subprocess, pathlib
out = pathlib.Path('biswa_ai_test.mp4')
cmd = [
    'python','third_party/Wan2.2/generate.py',
    '--task','ti2v-5B','--size','1280*704',
    '--ckpt_dir','models/Wan2.2-TI2V-5B',
    '--offload_model','True','--convert_model_dtype','--t5_cpu',
    '--image',INPUT,'--prompt',PROMPT,
    '--save_file',str(out), '--base_seed','42'
]
subprocess.run(cmd, check=True)
print('Created:', out)
```
